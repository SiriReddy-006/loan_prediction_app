```python
# Load datasets

df1 = pd.read_csv("loan_prediction.csv")

df2 = pd.read_csv("loan_status.csv")
```

```python
df1['Loan_Status'] = df2['Loan_Status']

df = df1

print(df.columns)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

columns = [
    'Gender',
    'Married',
    'Education',
    'Self_Employed',
    'Property_Area',
    'Loan_Status'
]

for col in columns:
    df[col] = encoder.fit_transform(df[col])

print(df.head())
```

```python
# Check Null Values

print(df.isnull().sum())
```

```python
# Fill Null Values

df.fillna(df.mode().iloc[0], inplace=True)

print(df.isnull().sum())
```

```python
# Separate Independent and Dependent Variables

x = df.drop(['Loan_ID', 'Loan_Status'], axis=1)

y = df['Loan_Status']

print(x.head())

print(y.head())
```

```python
# Convert Dependents Column

x['Dependents'] = x['Dependents'].replace('3+', 3)

x['Dependents'] = x['Dependents'].astype(int)
```

```python
# Handle Outliers

import numpy as np

numeric_columns = x.select_dtypes(include=np.number).columns

for col in numeric_columns:

    Q1 = x[col].quantile(0.25)
    Q3 = x[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    x = x[(x[col] >= lower) & (x[col] <= upper)]

# Adjust y values

y = y[x.index]

print(x.shape)

print(y.shape)
```

```python
# Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)

print(x_scaled)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training Data:", len(x_train))
print("Testing Data:", len(x_test))
```

```python
# Create Logistic Regression Model

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=5000)

model.fit(x_train, y_train)

print("Model Trained Successfully")
```

```python
# Predict Output

y_pred = model.predict(x_test)

print(y_pred)
```

```python
# Compare Actual and Predicted Values

comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

print(comparison)
```

```python
# Accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
# Confusion Matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Precision, Recall, F1 Score

from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

precision = precision_score(y_test, y_pred)

recall = recall_score(y_test, y_pred)

f1 = f1_score(y_test, y_pred)

print("Precision:", precision)

print("Recall:", recall)

print("F1 Score:", f1)
```

```python
# Classification Report

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)

print(report)
```
